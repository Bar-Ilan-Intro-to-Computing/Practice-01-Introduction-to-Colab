## 1) Measuring running time with `time.perf_counter()`

We start with a direct measurement of running time.


```python
import time
import matplotlib.pyplot as plt
```

```python
def stopper(n):
    t0 = time.perf_counter()
    i = 0
    while i < n:
        i += 1
    t1 = time.perf_counter()
    return t1 - t0
```

## 2) Linear time

The next two functions both go through a list, but not in exactly the same way.


```python
def linear(L):
    for i in range(len(L)):
        pass

def linear_too(L):
    for i in range(0, len(L), 2):
        pass
```

### Quick discussion

- `linear(L)` touches every element.
- `linear_too(L)` touches about half of the elements.

So one of them should be faster.

But do they belong to the same complexity class?


```python
def measure_runtime(func, arg, repeats=3):
    times = []
    for _ in range(repeats):
        t0 = time.perf_counter()
        func(arg)
        t1 = time.perf_counter()
        times.append(t1 - t0)
    return sum(times) / len(times)
```

```python
sizes = [10_000, 20_000, 40_000, 80_000, 160_000]
linear_times = []
linear_too_times = []

for n in sizes:
    L = list(range(n))
    linear_times.append(measure_runtime(linear, L))
    linear_too_times.append(measure_runtime(linear_too, L))

print("sizes:", sizes)
print("linear:", linear_times)
print("linear_too:", linear_too_times)
```

### Graph — linear growth

The graph below is useful because it shows **shape**, not just single numbers.


```python
plt.figure(figsize=(7,5))
plt.plot(sizes, linear_times, marker='o', label='linear')
plt.plot(sizes, linear_too_times, marker='o', label='linear_too')
plt.xlabel("input size (n)")
plt.ylabel("average running time (seconds)")
plt.title("Linear vs. linear_too")
plt.legend()
plt.grid(True)
plt.show()
```

## 4) Quadratic time

Now we move to nested loops.


```python
def quadratic(L):
    for i in range(len(L)):
        for j in range(len(L)):
            pass

def quadratic_too(L):
    for i in range(len(L)):
        for j in range(i, len(L)):
            pass
```

## 7) Binary search and logarithmic time

Binary search works on a **sorted** list.

Each step cuts the search space roughly in half.


```python
def binary_search(L, x):
    left, right = 0, len(L) - 1
    while left <= right:
        mid = (left + right) // 2
        if L[mid] == x:
            return True
        elif L[mid] > x:
            right = mid - 1
        else:
            left = mid + 1
    return False
```

### Task 7a — tracking the algorithm

Complete the function below so it prints the value of `mid` in each iteration.


```python
def binary_search_trace(L, x):
    left, right = 0, len(L) - 1
    while left <= right:
        mid = (left + right) // 2
        print("mid index:", mid)
        if L[mid] == x:
            return True
        elif L[mid] > x:
            right = mid - 1
        else:
            left = mid + 1
    return False
```

```python
L = [1, 3, 4, 5, 9, 12, 13, 26, 29, 32]

print("x = 3")
print(binary_search_trace(L, 3))
print()

print("x = 6")
print(binary_search_trace(L, 6))
print()

print("x = 33")
print(binary_search_trace(L, 33))
```

### Counting iterations in binary search


```python
def binary_search_steps(n):
    left, right = 0, n - 1
    steps = 0
    while left <= right:
        mid = (left + right) // 2
        steps += 1
        # force the "not found and go right" case just to count iterations
        left = mid + 1
    return steps

for n in [10, 100, 1000, 2000, 10_000]:
    print("n =", n, "-> steps =", binary_search_steps(n))
```

## 9) Optional challenge

The chapter also presents a naïve repeated-substring idea with nested loops.
Here is a simplified version.


```python
def repeating_substring_naive1(st, k):
    for i in range(len(st) - k + 1):
        for j in range(i + 1, len(st) - k + 1):
            if st[i:i+k] == st[j:j+k]:
                return st[i:i+k]
    return None
```
