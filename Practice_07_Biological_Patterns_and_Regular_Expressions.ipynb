# Part 4 – Finding All Matches

## `findall()`

````python
matches = re.findall("[Hh]ello", "Hello world, hello Python!")
print(matches)
````

## `finditer()`

````python
matches = re.finditer("[Hh]ello", "Hello dogs, hello cats!")

for m in matches:
    print(m.group(), m.span())
````

# Part 11 – Compiled Patterns
When we use the same regex many times,
we can “compile” it in advance using:

```python
re.compile()
```
Python stores the pattern in a more efficient form in memory,
so future searches become faster.

Without compile()
Every time Python:
* reads the regex
* analyzes it again
* then performs the search

With compile()
The regex is prepared once and reused many times.
This is especially useful when:
* searching the same pattern in many sequences
* working with large DNA/protein datasets
* running loops with repeated searches

````python
#Example
pattern = re.compile("[Hh]ello")

print(pattern.findall("Hello dogs, hello cats!"))
print(pattern.findall("Hello world"))
````
