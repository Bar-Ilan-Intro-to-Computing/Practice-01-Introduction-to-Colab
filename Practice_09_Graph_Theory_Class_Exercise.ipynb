# Part 0 – Helper Functions

The following functions will help us draw small graphs and inspect their structure.
You do not need to change this code.

````python
import random
import itertools
import math
import matplotlib.pyplot as plt
````

````python
def draw_graph(nodes, edges, positions=None, directed=False, title="Graph"):
    """Draw a small graph using matplotlib."""
    if positions is None:
        positions = {}
        n = len(nodes)
        for idx, node in enumerate(nodes):
            angle = 2 * math.pi * idx / n
            positions[node] = (math.cos(angle), math.sin(angle))

    plt.figure(figsize=(5, 4))

    for edge in edges:
        u, v = edge[0], edge[1]
        x1, y1 = positions[u]
        x2, y2 = positions[v]
        if directed:
            plt.arrow(x1, y1, x2-x1, y2-y1,
                      length_includes_head=True,
                      head_width=0.04,
                      alpha=0.6)
        else:
            plt.plot([x1, x2], [y1, y2], alpha=0.6)

    for node in nodes:
        x, y = positions[node]
        plt.scatter(x, y, s=600)
        plt.text(x, y, str(node), ha='center', va='center', color='white', fontsize=12)

    plt.title(title)
    plt.axis('off')
    plt.show()
````

````python
# Example graph

nodes = [0,1,2,3]
edges = [(0,1),(0,2),(1,2),(0,3)]

draw_graph(nodes, edges,
           title="Example Graph")
````
