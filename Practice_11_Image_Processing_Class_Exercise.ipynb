````python
from PIL import Image, ImageDraw
import matplotlib.pyplot as plt
````

````python
def show_image(im, title="Image"):
    """Display a PIL image using matplotlib."""
    plt.figure(figsize=(5, 4))
    plt.imshow(im, cmap="gray", vmin=0, vmax=255)
    plt.title(title)
    plt.axis("off")
    plt.show()


def show_images(images, titles=None):
    """Display several PIL images side by side."""
    if titles is None:
        titles = [f"Image {i+1}" for i in range(len(images))]

    plt.figure(figsize=(4 * len(images), 4))
    for i, im in enumerate(images):
        plt.subplot(1, len(images), i + 1)
        plt.imshow(im, cmap="gray", vmin=0, vmax=255)
        plt.title(titles[i])
        plt.axis("off")
    plt.show()
````

# Part 1 – Creating a Practice Image

Instead of downloading an external image, we will create a small grayscale image directly in Python.

This image will contain:

- a smooth gradient
- bright circles
- dark rectangles
- different gray levels

This makes it useful for testing image-processing functions.


````python
def create_practice_image(width=220, height=150):
    """Create a grayscale practice image with several visible objects."""

    im = Image.new(mode="L", size=(width, height), color=0)
    mat = im.load()

    # Horizontal gradient background
    for x in range(width):
        for y in range(height):
            mat[x, y] = int(40 + 180 * x / (width - 1))

    draw = ImageDraw.Draw(im)

    # Bright biological-like objects
    draw.ellipse((30, 30, 75, 75), fill=245)
    draw.ellipse((90, 45, 135, 90), fill=210)
    draw.ellipse((155, 25, 200, 70), fill=250)

    # Dark structures
    draw.rectangle((40, 105, 95, 130), fill=25)
    draw.rectangle((135, 100, 195, 125), fill=70)

    # Medium gray line
    draw.line((0, height - 1, width - 1, 0), fill=130, width=3)

    return im

practice_im = create_practice_image()
show_image(practice_im, "Practice grayscale image")
print("Image size:", practice_im.size)
````
